```python
import cv2
import os

os.makedirs("frames", exist_ok=True)

cap = cv2.VideoCapture("traffic_video.mp4")

count = 0

while True:

    ret, frame = cap.read()

    if not ret:
        break

    cv2.imwrite(
        f"frames/frame_{count:04d}.jpg",
        frame
    )

    count += 1

cap.release()

print("Frames Extracted:", count)
```

```python
import os

folders = [

    "dataset/images/train",
    "dataset/images/val",
    "dataset/images/test",

    "dataset/labels/train",
    "dataset/labels/val",
    "dataset/labels/test"
]

for folder in folders:
    os.makedirs(folder, exist_ok=True)

print("Folders Created")
```

```python
import os
import shutil

images = sorted(os.listdir("frames"))

train = images[:100]
val = images[100:140]
test = images[140:]

for img in train:

    shutil.copy(
        f"frames/{img}",
        f"dataset/images/train/{img}"
    )

for img in val:

    shutil.copy(
        f"frames/{img}",
        f"dataset/images/val/{img}"
    )

for img in test:

    shutil.copy(
        f"frames/{img}",
        f"dataset/images/test/{img}"
    )

print("Dataset Split Complete")
```
