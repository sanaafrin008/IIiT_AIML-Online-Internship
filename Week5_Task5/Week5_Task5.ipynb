```python
import cv2
import os

image_folder = "/content/runs/detect/predict"

images = sorted([
    img for img in os.listdir(image_folder)
    if img.endswith(".jpg")
])

frame = cv2.imread(
    os.path.join(image_folder, images[0])
)

height, width, layers = frame.shape

video = cv2.VideoWriter(
    'demo_video.mp4',
    cv2.VideoWriter_fourcc(*'mp4v'),
    1,
    (width, height)
)

for image in images:
    video.write(
        cv2.imread(
            os.path.join(image_folder, image)
        )
    )

video.release()

print("Video created")
```
