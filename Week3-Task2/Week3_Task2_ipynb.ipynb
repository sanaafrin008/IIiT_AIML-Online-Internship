```python
import cv2
import os

os.makedirs("frames", exist_ok=True)

cap = cv2.VideoCapture("raw_video.mp4")

count = 0

while True:
    ret, frame = cap.read()

    if not ret:
        break

    cv2.imwrite(
        f"frames/frame_{count:04d}.jpg",
        frame
    )

    count += 1

cap.release()

print("Frames:", count)
```

```python
from ultralytics import YOLO

det_model = YOLO("yolov8n.pt")

det_model.predict(
    source="frames",
    save=True,
    project="runs",
    name="detect"
)
```

```python
from ultralytics import YOLO

seg_model = YOLO("yolov8n-seg.pt")

seg_model.predict(
    source="frames",
    save=True,
    project="runs",
    name="segment"
)
```

```python
import cv2
import os

folder = "/content/runs/detect/runs/detect"

images = sorted([
    x for x in os.listdir(folder)
    if x.endswith(".jpg")
])

frame = cv2.imread(
    os.path.join(folder, images[0])
)

h,w,_ = frame.shape

video = cv2.VideoWriter(
    "detection_video.mp4",
    cv2.VideoWriter_fourcc(*'mp4v'),
    30,
    (w,h)
)

for img in images:

    video.write(
        cv2.imread(
            os.path.join(folder,img)
        )
    )

video.release()

print("Detection Video Ready")
```

```python
import cv2
import os

folder = "/content/runs/segment/runs/segment"

images = sorted([
    x for x in os.listdir(folder)
    if x.endswith(".jpg")
])

frame = cv2.imread(
    os.path.join(folder, images[0])
)

h,w,_ = frame.shape

video = cv2.VideoWriter(
    "segmentation_video.mp4",
    cv2.VideoWriter_fourcc(*'mp4v'),
    30,
    (w,h)
)

for img in images:

    video.write(
        cv2.imread(
            os.path.join(folder,img)
        )
    )

video.release()

print("Segmentation Video Ready")
```

```python
from moviepy.editor import *

video = VideoFileClip(
    "no_audio.mp4"
)

audio = AudioFileClip(
    "music.mp3"
)

audio = audio.subclip(
    0,
    min(
        video.duration,
        audio.duration
    )
)

final = video.set_audio(audio)

final.write_videofile(
    "final_output.mp4",
    codec="libx264",
    audio_codec="aac"
)
```
